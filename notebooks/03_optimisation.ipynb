```python
import pandas as pd
import joblib
import sys
from pathlib import Path
PROJECT_ROOT = Path.cwd().parent
if not (PROJECT_ROOT / "src").is_dir():
    PROJECT_ROOT = Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.preprocessing import encode_categorical, split_data
from src.tune_models import tune_random_forest, tune_xgboost, evaluate_model

df = pd.read_csv("../data/processed/cars-clean.csv")
df_encoded = encode_categorical(df, ['fuel', 'seller_type', 'transmission'])

X_train, X_test, y_train, y_test = split_data(df_encoded)
numeric_cols = ['year', 'km_driven', 'owner']

rf_search = tune_random_forest(X_train, y_train, numeric_cols, n_iter=20, cv=5)

print("Meilleurs paramètres RF :", rf_search.best_params_)
print("Meilleur score CV (RMSE) :", -rf_search.best_score_)
```

```python
xgb_search = tune_xgboost(X_train, y_train, numeric_cols, n_iter=20, cv=5)

print("Meilleurs paramètres XGB :", xgb_search.best_params_)
print("Meilleur score CV (RMSE) :", -xgb_search.best_score_)
```
